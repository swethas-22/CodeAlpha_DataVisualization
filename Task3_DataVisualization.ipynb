```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set a clean visual style for all charts
sns.set_theme(style="whitegrid")

# Load the built-in Restaurant Tips dataset
df = sns.load_dataset('tips')

# View the first 5 rows
df.head()
```

```python
# 1. Bar Chart: Average Total Bill by Day of the Week
plt.figure(figsize=(7, 4))
sns.barplot(data=df, x='day', y='total_bill', palette='viridis', errorbar=None)
plt.title('Average Total Bill by Day of the Week', fontsize=14)
plt.xlabel('Day of the Week')
plt.ylabel('Average Bill ($)')
plt.show()

# 2. Scatter Plot: Relationship Between Total Bill and Tip Amount
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time', style='time', s=80)
plt.title('Total Bill vs. Tip Amount (Lunch vs. Dinner)', fontsize=14)
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()

# 3. Pie Chart: Proportion of Customers by Meal Time
meal_counts = df['time'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(meal_counts, labels=meal_counts.index, autopct='%1.1f%%', colors=['#66b3ff','#ff9999'], startangle=90)
plt.title('Proportion of Customers: Dinner vs. Lunch', fontsize=14)
plt.show()

# 4. Correlation Heatmap: Numerical Variables
plt.figure(figsize=(6, 4))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Restaurant Metrics', fontsize=14)
plt.show()
```

Weekend Revenue Peak: Average bills are highest on Saturdays and Sundays, indicating the need for more weekend staff.

Tipping Behavior: There is a strong positive correlation (0.68) between total bill size and tip amount.

Shift Traffic: Dinner accounts for 72.1% of customer traffic compared to 27.9% for lunch.
